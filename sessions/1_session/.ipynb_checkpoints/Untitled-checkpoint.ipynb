# Testing logistic regression compared to Stata

```python
import pandas as pd
import statsmodels.api as sm
```

```python
data = pd.read_stata("https://stats.idre.ucla.edu/stat/stata/dae/binary.dta")
```

```python
data.head()
```

```python
ranks = pd.get_dummies(data['rank'])
```

```python
x = pd.concat([data[['gre', 'gpa']],ranks], axis=1)
```

```python
x = sm.add_constant(x)

y = data['admit']

log_reg_statsmodels = sm.Logit(y, x).fit()

log_reg_statsmodels.summary()
```
